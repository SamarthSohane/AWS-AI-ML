# Fraud/Anomaly Detection with Drift Monitoring

A reproducible time-aware benchmark using the Kaggle/ULB credit-card fraud data. The notebook performs data audit and EDA, class-weighted supervised modeling, normal-only Isolation Forest, validation-only cost threshold selection, held-out evaluation, PSI/KS and score-drift monitoring, a synthetic concept-drift stress test, and a clearly labeled retraining prototype.

> **Research artifact, not a production payment control.** Data and cost assumptions are illustrative; use institution-specific validation before operational use.

## Experimental protocol

The raw data contain a tiny positive class. Accuracy is therefore not a useful primary metric. Average Precision (PR-AUC) is emphasized alongside ROC-AUC, precision, recall, F1, confusion counts, false-positive/negative rates, and expected cost. Rows are time-sorted into 60% train, 20% validation, and 20% test windows. Transformations and models are fit on training rows only. Thresholds are optimized on validation cost and frozen before test evaluation.

Exact duplicate records are audited and removed before splitting. The Isolation Forest objective receives only legitimate training features. A separate synthetic concept-shift experiment changes labels while leaving features untouched; it must not be confused with observed dataset outcomes.

## Threshold and business-cost analysis

The cost assumption is configurable: `$5` per false positive (illustrative review/friction cost) and `$90.37` per false negative (80% of mean training-window fraud amount). The HistGradientBoosting operating threshold minimizes this expected cost on validation scores. Class weighting changes score calibration, so the selected numerical threshold is not a calibrated probability. The test rows are not involved in threshold selection.

````python
final_name = json.loads((RESULTS / "selected_thresholds.json").read_text())["supervised_model_selected_by_validation_pr_auc"]
threshold_table = pd.read_csv(RESULTS / "threshold_analysis.csv")
final_display_name = "HistGradientBoosting (class-weighted)" if final_name == "hist_gradient_boosting" else "Logistic Regression (class-weighted baseline)"
display(threshold_table.loc[threshold_table.model == final_display_name, ["threshold", "precision", "recall", "f1", "false_positives", "false_negatives", "estimated_total_cost", "selected_cost_threshold"]])
display(Image(filename=str(FIGURES / "threshold_precision_recall.png")))
display(Image(filename=str(FIGURES / "threshold_expected_cost.png")))
````

## Drift monitoring

Feature distributions are compared between training/reference and the later test/current window using quantile-bin PSI and two-sample KS tests with Benjamini–Hochberg correction. The monotonically increasing `Time` counter is excluded from actionable feature alerts because its non-overlapping distributions are guaranteed by the split. Score distributions use out-of-sample validation scores as reference and test scores as current. PSI bands (<0.10, 0.10–0.25, ≥0.25) are practical heuristics, not universal thresholds.

````python
drift = pd.read_csv(RESULTS / "drift_report.csv")
score_drift = pd.read_csv(RESULTS / "model_score_drift.csv")
alert = json.loads((RESULTS / "drift_alert.json").read_text())
print((RESULTS / "drift_monitoring_report.txt").read_text())
display(drift.head(12))
display(score_drift)
print("Alert JSON:", alert)
````

## Synthetic concept drift and prototype retraining

The stress test leaves every feature unchanged and artificially raises fraud incidence in a low-score feature tail, changing `P(Y|X)`. It is explicitly synthetic, not an observed property of the Kaggle data. Feature PSI alone cannot detect this label-only intervention; a delayed-label performance monitor detects the resulting AP/recall drop. The optional prototype adapts on 80% of the synthetic validation window, calibrates on the remaining 20%, and evaluates on the separate synthetic test window. Its outputs are for process demonstration only.

````python
injection = json.loads((RESULTS / "concept_drift_injection.json").read_text())
performance_alert = json.loads((RESULTS / "concept_drift_performance_alert.json").read_text())
print("Region and injection:")
print(json.dumps(injection, indent=2))
print("Delayed-label performance alert:")
print(json.dumps(performance_alert, indent=2))
display(pd.read_csv(RESULTS / "concept_drift_performance.csv"))
display(pd.read_csv(RESULTS / "retraining_comparison.csv"))
display(Image(filename=str(FIGURES / "concept_drift_performance.png")))
display(Image(filename=str(FIGURES / "retraining_performance_comparison.png")))
````
